### Phase 2: DAG Builder

In this phase, a Directed Acyclic Graph (DAG) is constructed to represent the causal relationship between the treatment and outcome variables.

The graph is intentionally kept simple to reflect the direct relationship between the variables without introducing additional assumptions.

```python
import pickle
with open("../artifacts/phase1_data_understanding/data_schema.pkl", "rb") as f:
    SCHEMA = pickle.load(f)
```

```python
class DAGBuilder:
    def __init__(self, treatment, outcome, confounders, extra_edges = None):
        self.treatment = treatment
        self.outcome = outcome
        self.confounders = confounders
        self.extra_edges = extra_edges if extra_edges else []
    def build_graph(self):
        edges = []
        edges.append(f"{self.treatment} -> {self.outcome};")
        for var in self.confounders:
            edges.append(f"{var} -> {self.treatment};")
            edges.append(f"{var} -> {self.outcome};")
        for edge in self.extra_edges:
            edges.append(edge)
        graph = "digraph {\n"
        for e in edges:
            graph += f"    {e}\n"
        graph += "}"
        return graph
```

### DAG Construction

A directed edge is created from the treatment variable to the outcome variable.

This represents the assumption that study time has a direct influence on student performance.

```python
builder = DAGBuilder(treatment = "studytime", outcome = "G3", confounders = ["absences", "health", "failures"])
causal_graph = builder.build_graph()
print(causal_graph)
```

### Variable Selection

The treatment variable is defined as **studytime**, representing the amount of time a student spends studying.

The outcome variable is defined as **G3**, which represents the final grade.

These variables are selected based on their direct relationship in the dataset.

### Graph Representation

The graph shows a single directional relationship from studytime to G3.

This confirms that the model considers study time as a causal factor influencing the final grade.

```python
with open("../artifacts/phase2_dag_generation/causal_dag.pkl", "wb") as f:
    pickle.dump(causal_graph, f)
```

### Conclusion

The DAG defines a simple causal structure that will be used in the next phase for estimating the causal effect.
